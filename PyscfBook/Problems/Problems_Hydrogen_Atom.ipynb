```python
psi_100 = Psi_nlm(1, 0, 0, r, theta, phi, zeta)
psi_100
```

Нормування:
$$
\langle\psi_{100}|\psi_{100}\rangle = 4\pi\int_0^\infty \psi_{100}^*\psi_{100}\,r^2\,dr

```python
norm_100 = integrate(conjugate(psi_100) * psi_100 * 4 * pi * r**2, (r, 0, oo))
norm_100
```

Середні $\langle U\rangle$, $\langle T\rangle$, $\langle E\rangle$ та звірка з точною формулою $E_n=-Z^2/(2n^2)$:

```python
U = -Z / r
U_100 = integrate(conjugate(psi_100) * U * psi_100 * 4 * pi * r**2, (r, 0, oo))

T_psi_100 = -Rational(1, 2) * (1 / r**2) * diff(r**2 * diff(psi_100, r), r)
T_100 = integrate(conjugate(psi_100) * T_psi_100 * 4 * pi * r**2, (r, 0, oo))

E_100 = (T_100 + U_100).subs(zeta, Z)
E_100_exact = E_nl(1, Z)

print("⟨U⟩ =", U_100, " ⟨T⟩ =", T_100)
print("⟨E⟩ (варіаційно) =", E_100, "   E_1 (точна формула) =", E_100_exact)
```

Теорія дає загальну формулу середнього радіуса для довільних $n,l$:

(в атомних одиницях $a_0=1$). Ваше завдання — обчислити $\langle r\rangle$ **прямим інтегруванням** для кількох станів через радіальну функцію $R_{nl}(r,Z)$ з `sympy.physics.hydrogen`, а потім звірити з формулою вище.
