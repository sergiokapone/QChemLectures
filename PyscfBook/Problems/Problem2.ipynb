# Задача (варіаційний метод для атома гелію)

Розглянути атом гелію (ядерний заряд $Z=2$) у нерелятивістському наближенні, в атомних одиницях.  
В якості **пробної** хвильової функції основного стану візьміть  добуток двох 1s-орбіталей (факторизація Хартрі)
з ефективним зарядом $\zeta$ (варіаційний параметр):

$$
\Psi_{\rm trial}(\mathbf r_1,\mathbf r_2;\zeta)=N(\zeta)\,e^{-\zeta(r_1+r_2)},\qquad r_i=|\mathbf r_i|.
$$

---

### Завдання:

1. **Нормування.**  
   Знайти константу нормування $N(\zeta)$ використовуючи sympy.

2. **Гамільтоніан.**  
   Записати повний гамільтоніан двоелектронної системи в атомних одиницях:
   $$
   H=-\tfrac12\nabla_1^2-\tfrac12\nabla_2^2 -\frac{Z}{r_1}-\frac{Z}{r_2} +\frac{1}{r_{12}},
   \qquad r_{12}=|\mathbf r_1-\mathbf r_2|.
   $$

3. **Очікування енергії.**  
   Обчислити на пробній функції:
   - кінетичну енергію $\langle T\rangle$,
   - енергію взаємодії електрон–ядро $\langle V_\text{en}\rangle$,
   - кулонівське відштовхування $\langle r_{12}^{-1}\rangle$. Тут можна скористаитись виразом з презентацій до лекції.

   Записати загальний вираз:
   $$
   E(\zeta)=\langle\Psi_{\rm trial}|H|\Psi_{\rm trial}\rangle.
   $$

4. **Мінімізація.**  
   Знайти значення $\zeta^\star$, яке мінімізує $E(\zeta)$,
   та відповідну енергію основного стану:
   $$
   E_{\rm var}=E(\zeta^\star).
   $$

5. **Числові значення.**  
   Навести $\zeta^\star$ та $E_{\rm var}$ у атомних одиницях і в електронвольтах
   (використати $1\ \text{a.u.}=27.211386\ \text{eV}$).
